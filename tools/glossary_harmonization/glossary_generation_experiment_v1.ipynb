    - HELIO Ontology: A solar event (CME) that involves a burst of plasma ejected into the interplanetary medium. CME's may be observed remotely relatively near the sun or in situ in the interplanetary medium. The latter type of observations are often referred to as Interplanetary CME's (ICME's).

At their largest, CMEs can contain 10 billion tons of matter, and they can move at speeds of a million miles an hour. Semantic caution: “near-Earth” is fundamentally locational, while the physical process is magnetotail reconnection. I would therefore probably model:

magnetic reconnection
→ magnetotail reconnection
→ near-Earth magnetotail reconnection

rather than making near-Earth reconnection a fundamental physical subclass.

A useful hierarchy

I would not yet make these three sibling concepts without qualification:

Magnetic reconnection
│
├── Dayside magnetopause reconnection
│
└── Magnetotail reconnection
     │
     ├── Near-Earth magnetotail reconnection
     └── Distant-tail reconnection

This is much more ontologically defensible than treating dayside, near-Earth, and tail as three equivalent physical kinds of reconnection. Dayside/tail primarily identifies location/topology; near-Earth/distant-tail further specifies location within the tail.

2. Children of aurora

There is particularly strong support for distinguishing diffuse, monoenergetic, and broadband aurora on the basis of the precipitation/acceleration mechanism. Newell et al. explicitly construct a four-way classification: electron diffuse, ion diffuse, monoenergetic, and broadband.

Diffuse aurora
Source	Definition / description
Newell et al., JGR	Diffuse aurora results when particles precipitate without the acceleration events used to identify monoenergetic or broadband aurora; both electron and ion diffuse precipitation are included.

Candidate definition:

Diffuse aurora is auroral emission produced by relatively unaccelerated magnetospheric charged particles precipitating into the upper atmosphere, producing a broad and generally diffuse region of auroral luminosity rather than discrete accelerated auroral structures.

Important: “unaccelerated” is useful in the Newell classification, but I would not make it an absolute physical assertion in the ontology. Better to encode the distinction as a relationship to particle acceleration mechanism.

Broadband aurora
Source	Definition / description
Newell et al., JGR	Broadband acceleration is electron acceleration occurring over a wide range of energies, associated with dispersive Alfvén waves; it is also called wave aurora.

Newell et al.

Broadband aurora is auroral emission associated with electron precipitation produced by acceleration over a broad range of particle energies, commonly associated with dispersive Alfvén-wave acceleration.

Important semantic issue: This is really a precipitation/acceleration classification, not simply an optical morphology. I would therefore connect:

Proton aurora

This one needs a geographic qualifier because “proton aurora” is used for different planetary contexts.

For Earth:

Source	Definition / description

Proton aurora is auroral emission produced by precipitating magnetospheric protons entering the upper atmosphere and depositing energy that produces atmospheric emissions.

This is an excellent harmonization case: the common semantic core is auroral emission associated with proton/ion precipitation, but the particle-entry mechanism differs between planetary environments. I would not bake Earth's mechanism into the top-level definition.
